## Empresa ficticia

**NovaConnect Solutions** es una empresa ficticia especializada en servicios de atención al cliente y soporte multicanal para una compañía nacional de comercio electrónico dedicada a la venta de productos tecnológicos y del hogar.

La empresa gestiona todas las interacciones con los clientes posteriores a la compra, así como parte del proceso comercial, ofreciendo soporte a través de diferentes canales de comunicación.

---

# Objetivo del proyecto

El objetivo es desarrollar un sistema de análisis y predicción que permita estimar el volumen de contactos futuros y optimizar la planificación de personal mediante técnicas de análisis de datos y Machine Learning.

El proyecto simula un entorno real de Workforce Management (WFM), donde una planificación eficiente resulta fundamental para mantener los niveles de servicio y optimizar los recursos disponibles.

---

# Canales de atención

NovaConnect ofrece soporte mediante tres canales principales:

- 📞 Teléfono
- 💬 Chat
- 📧 Email

Cada canal presenta diferentes características en cuanto a volumen de contactos y duración media de las interacciones.

---

# Departamentos

## Atención al Cliente (ATC)

Responsable de gestionar:

- Información sobre pedidos.
- Cambios y devoluciones.
- Consultas generales.
- Incidencias sencillas.

Es el departamento con mayor volumen de contactos.

---

## Soporte Técnico

Gestiona incidencias relacionadas con:

- Configuración de productos.
- Garantías.
- Problemas técnicos.
- Resolución de averías.

Presenta un volumen menor de contactos, pero una duración media significativamente superior.

---

## Ventas

Encargado de:

- Consultas comerciales.
- Promociones.
- Venta cruzada.
- Información sobre nuevos productos.

Es el departamento con menor volumen y menor tiempo medio de atención.

---

# Horario de atención

La empresa presta servicio:

- **Lunes a domingo**
- **08:00 a 22:00 horas**

Los mayores picos de actividad suelen concentrarse entre:

- 10:00 - 14:00
- 17:00 - 19:00

---

# Organización

La empresa cuenta aproximadamente con:

- 120 agentes.
- Organización por turnos.
- Planificación semanal.
- Un equipo de Workforce Management responsable del dimensionamiento de personal.

---

# Objetivos operativos

Los principales indicadores operativos son:

- SLA objetivo del **90 %**.
- Reducir la tasa de abandono.
- Mantener una ocupación equilibrada de los agentes.
- Optimizar la planificación de recursos.
- Mejorar la experiencia del cliente.

---

# Campañas comerciales

Durante determinados periodos del año se producen incrementos significativos del volumen de contactos debido a campañas comerciales:

- Rebajas
- Black Friday
- Navidad

Estas campañas afectan tanto al número de contactos como al tiempo medio de atención (AHT).

---

# Incidencias operativas

De forma ocasional pueden producirse eventos extraordinarios como:

- Caídas de sistemas.
- Incidencias en la página web.
- Retrasos logísticos.
- Problemas con proveedores.

Estas situaciones generan aumentos puntuales del volumen de contactos y pueden afectar al cumplimiento del SLA.

---

# Dataset

Para el desarrollo del proyecto se utilizará un **dataset sintético** diseñado para simular la operativa diaria de un contact center.

El dataset contiene información sobre:

- Contactos individuales.
- Planificación de agentes.
- Calendario operativo.
- Campañas comerciales.

Los datos han sido generados siguiendo reglas de negocio realistas con el objetivo de reproducir el comportamiento habitual de una operación de atención al cliente.
